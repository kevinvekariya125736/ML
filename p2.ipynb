```python
import pandas as pd

df = pd.read_csv("Dataset_pandas_exercisecsv__2026_10_14_08_08_12.csv")
```

```python

print("FIRST FIVE ROWS")
print(df.head())
print("LAST FIVE ROWS")
print(df.tail())
```

```python
df = df.replace(["?", "n.a", "NaN", "nan"], pd.NA)
df.to_csv("cleaned_dataset.csv", index=False)
print("Cleaned data saved to cleaned_dataset.csv")
```

```python
most_expensive = df.loc[df["price"].idxmax()]
print("Most expensive company:", most_expensive["company"])
print("Price:", most_expensive["price"])
```

```python
print(df[df["company"].str.lower() == "toyota"])
```

```python
print(df["company"].value_counts().sort_index())
```

```python
valid_prices = df.dropna(subset=["price"])
highest_each = valid_prices.loc[valid_prices.groupby("company")["price"].idxmax()]
print(highest_each[["company", "price"]].sort_values("company"))
```

```python
print(df.groupby("company")["average-mileage"].mean().round(2))
```

```python
print(df.sort_values("price", na_position="last"))
```

```python
GermanCars = pd.DataFrame({
    "Company": ["Ford", "Mercedes", "BMV", "Audi"],
    "Price": [23845, 171995, 135925, 71400]
})
japaneseCars = pd.DataFrame({
    "Company": ["Toyota", "Honda", "Nissan", "Mitsubishi"],
    "Price": [29995, 23600, 61500, 58900]
})
all_cars = pd.concat([GermanCars, japaneseCars], ignore_index=True)
print(all_cars)
```

```python
Car_Price = pd.DataFrame({
    "Company": ["Toyota", "Honda", "BMV", "Audi"],
    "Price": [23845, 17995, 135925, 71400]
})
car_Horsepower = pd.DataFrame({
    "Company": ["Toyota", "Honda", "BMV", "Audi"],
    "horsepower": [141, 80, 182, 160]
})
merged_cars = pd.merge(Car_Price, car_Horsepower, on="Company")
print(merged_cars)
```
