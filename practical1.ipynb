```python
import pandas as pd

df = pd.read_csv("auto-mpg.csv")

print("Numeric Attributes:")
print(df.select_dtypes(include='number').columns.tolist())

print("Categorical Attributes:")
print(df.select_dtypes(include='object').columns.tolist())
```

4th


```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

df = pd.read_csv('auto-mpg.csv')
df[df.select_dtypes('number').columns] = MinMaxScaler().fit_transform(df.select_dtypes('number'))
print(df)
```

5th

```python
df = pd.get_dummies(df, columns=df.select_dtypes('object').columns)
print(df)
```

6th

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

X = df.select_dtypes('number').dropna()

X = StandardScaler().fit_transform(X)
print(PCA(n_components=2).fit_transform(X))
```
