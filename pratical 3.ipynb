```python
import pandas as pd
```

```python
data = pd.read_csv("supervised_learning.csv")
data
```

```python
X = data.iloc[:,:-1]
Y = data.iloc[:,-1]
X
Y
```

```python
X['Age'] = X['Age'].map({'Youth': 1, 'Middle': 2, 'Senior': 3})
X['Income'] = X['Income'].map({'High': 1, 'Medium': 2, 'Low': 3})
X['Student'] = X['Student'].map({'No': 0, 'Yes': 1})
X['Creditrating'] = X['Creditrating'].map({'Fair': 0, 'Excellent': 1})
X
```

```python
from sklearn.naive_bayes import GaussianNB
nb=GaussianNB()
nb.fit(X,Y)
print(nb.predict([[3,3,1,1]]))
print(nb.predict([[1,1,0,0]]))
print(nb.predict([[1,1,1,1]]))
```
